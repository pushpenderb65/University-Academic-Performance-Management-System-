```python
import random
from faker import Faker
import pandas as pd
from sqlalchemy import create_engine
import warnings
warnings.filterwarnings("ignore")
```

```python
fake = Faker('en_IN')

# MySQL Connection String: mysql+pymysql://username:password@localhost:3306/database_name
engine = create_engine('mysql+pymysql://root:2003@localhost:3306/university_db')
```

```python
# 1. Populate Departments
depts = ['Computer Science', 'Data Science', 'Electronics', 'Mechanical', 'Civil']
dept_df = pd.DataFrame([{
    'dept_name': d,
    'building_name': f'Block-{chr(65+i)}',
    'budget': round(random.uniform(500000, 2000000), 2)
} for i, d in enumerate(depts)])
dept_df.to_sql('Departments', engine, if_exists='append', index=False)
```

```python
# 2. Populate Instructors (25 rows)
instructors = [{
    'first_name': fake.first_name(),
    'last_name': fake.last_name(),
    'email': fake.unique.email(),
    'dept_id': random.randint(1, 5),
    'hire_date': fake.date_between(start_date='-8y', end_date='-1y'),
    'salary': round(random.uniform(50000, 150000), 2)
} for _ in range(25)]
pd.DataFrame(instructors).to_sql('Instructors', engine, if_exists='append', index=False)
```

```python
# 3. Populate Courses (15 rows)
courses_data = [
    ('CS101', 'Data Structures & Algorithms', 4, 1),
    ('CS102', 'Database Management Systems', 4, 1),
    ('DS101', 'Applied Machine Learning', 4, 2),
    ('DS102', 'Data Visualization with BI', 3, 2),
    ('EC101', 'Digital Signal Processing', 4, 3),
    ('EC102', 'Microcontrollers & IoT', 3, 3),
    ('ME101', 'Thermodynamics', 4, 4),
    ('ME102', 'Fluid Mechanics', 4, 4),
    ('CE101', 'Structural Analysis', 4, 5),
    ('CE102', 'Environmental Engineering', 3, 5),
]
course_df = pd.DataFrame(courses_data, columns=['course_code', 'course_title', 'credits', 'dept_id'])
course_df.to_sql('Courses', engine, if_exists='append', index=False)
```

```python
# 4. Populate Students (500 rows)
students = [{
    'first_name': fake.first_name(),
    'last_name': fake.last_name(),
    'email': fake.unique.email(),
    'dept_id': random.randint(1, 5),
    'enrollment_year': random.choice([2023, 2024, 2025, 2026]),
    'current_semester': random.randint(1, 8)
} for _ in range(500)]
pd.DataFrame(students).to_sql('Students', engine, if_exists='append', index=False)
```

```python



# 5. Populate Sections (10 sections)
sections = [{
    'course_id': random.randint(1, 10),
    'instructor_id': random.randint(1, 25),
    'semester': 'Odd',
    'academic_year': 2026,
    'room_number': f'LH-{random.randint(101, 305)}'
} for _ in range(10)]
pd.DataFrame(sections).to_sql('Sections', engine, if_exists='append', index=False)

```
