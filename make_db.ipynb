```python
import pandas as pd
from sqlalchemy import create_engine

# Aapka local MySQL connection
mysql_engine = create_engine('mysql+pymysql://root:2003@localhost:3306/university_db')

# SQLite database file create hogi
sqlite_engine = create_engine('sqlite:///university.db')

tables = ['Departments', 'Instructors', 'Courses', 'Students', 'Sections', 'Enrollments']

for table in tables:
    df = pd.read_sql(f"SELECT * FROM {table}", mysql_engine)
    df.to_sql(table, sqlite_engine, if_exists='replace', index=False)
    print(f"✅ {table} converted")

# student_gpa_report View ko bhi SQLite me save karein
view_df = pd.read_sql("SELECT * FROM student_gpa_report", mysql_engine)
view_df.to_sql('student_gpa_report', sqlite_engine, if_exists='replace', index=False)
print("✅ student_gpa_report converted")
```
